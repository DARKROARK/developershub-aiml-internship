## Step 1: Import Required Libraries

```python
# Core data handling
import pandas as pd
import numpy as np

# Visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Set a clean visual style for all plots
sns.set_theme(style='whitegrid', palette='Set2')
plt.rcParams['figure.figsize'] = (8, 5)
plt.rcParams['figure.dpi'] = 100

print('All libraries imported successfully!')
```

## Step 2: Load the Dataset

We load the Iris dataset directly using Seaborn's built-in dataset loader. This returns a pandas DataFrame.

```python
# Load the Iris dataset via seaborn
df = sns.load_dataset('iris')

print('Dataset loaded successfully!')
print(f'Shape of dataset: {df.shape}')  # (rows, columns)
print(f'Column names: {list(df.columns)}')
```

`.info()` tells us the data types, number of non-null entries, and memory usage — important for spotting missing values.

`.describe()` gives us count, mean, standard deviation, min/max, and quartile values for all numerical columns.

## Step 6: Data Visualization

### 6.1 Scatter Plot — Relationships Between Features

A **pair plot** (scatter matrix) shows pairwise relationships between all features, colored by species. This is the best single plot for understanding multi-feature separation.

```python
# Pair plot — scatter plots for all feature combinations
sns.pairplot(df, hue='species', markers=['o', 's', 'D'], plot_kws={'alpha': 0.7})
plt.suptitle('Pair Plot of Iris Dataset Features by Species', y=1.02, fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('scatter_pairplot.png', bbox_inches='tight', dpi=120)
plt.show()
print('Scatter pair plot saved as scatter_pairplot.png')
```

```python
# Focused scatter plot: Petal Length vs Petal Width (best separator)
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df,
    x='petal_length',
    y='petal_width',
    hue='species',
    style='species',
    s=80,
    alpha=0.8
)
plt.title('Petal Length vs Petal Width by Species', fontsize=13, fontweight='bold')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Petal Width (cm)')
plt.legend(title='Species')
plt.tight_layout()
plt.savefig('scatter_petal.png', dpi=120)
plt.show()
print('Petal scatter plot saved as scatter_petal.png')
```

**Insight:** Petal length and petal width show clear cluster separation between the three species — `setosa` is easily distinguishable, while `versicolor` and `virginica` slightly overlap.

### 6.2 Histograms — Feature Value Distributions

Histograms reveal the distribution shape of each feature — whether it's normal, skewed, or multimodal.

```python
# Histogram for each feature, broken down by species
features = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
axes = axes.flatten()

for i, feature in enumerate(features):
    for species in df['species'].unique():
        subset = df[df['species'] == species]
        axes[i].hist(subset[feature], bins=15, alpha=0.6, label=species, edgecolor='white')

    axes[i].set_title(f'Distribution of {feature.replace("_", " ").title()}', fontweight='bold')
    axes[i].set_xlabel(f'{feature.replace("_", " ").title()} (cm)')
    axes[i].set_ylabel('Frequency')
    axes[i].legend(title='Species')

plt.suptitle('Feature Distributions by Species', fontsize=14, fontweight='bold', y=1.01)
plt.tight_layout()
plt.savefig('histograms.png', bbox_inches='tight', dpi=120)
plt.show()
print('Histograms saved as histograms.png')
```

**Insight:** Petal length and petal width show multimodal distributions (multiple peaks) — one peak per species. This confirms these features are strong predictors for species classification. Sepal width shows the most overlap between species.

### 6.3 Box Plots — Identifying Outliers and Spread

Box plots show the median, interquartile range (IQR), and outliers for each feature per species.

```python
# Box plots for all 4 features grouped by species
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
axes = axes.flatten()

for i, feature in enumerate(features):
    sns.boxplot(
        data=df,
        x='species',
        y=feature,
        palette='Set2',
        ax=axes[i],
        width=0.5,
        flierprops=dict(marker='o', color='red', markersize=5)  # Outliers in red
    )
    axes[i].set_title(f'{feature.replace("_", " ").title()} by Species', fontweight='bold')
    axes[i].set_xlabel('Species')
    axes[i].set_ylabel(f'{feature.replace("_", " ").title()} (cm)')

plt.suptitle('Box Plots — Feature Distribution and Outliers by Species', fontsize=13, fontweight='bold', y=1.01)
plt.tight_layout()
plt.savefig('boxplots.png', bbox_inches='tight', dpi=120)
plt.show()
print('Box plots saved as boxplots.png')
```

```python
# Combined violin + box plot for richer view
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Sepal length violin
sns.violinplot(data=df, x='species', y='sepal_length', palette='Set2', ax=axes[0], inner='box')
axes[0].set_title('Sepal Length Distribution (Violin)', fontweight='bold')
axes[0].set_xlabel('Species')
axes[0].set_ylabel('Sepal Length (cm)')

# Petal length violin
sns.violinplot(data=df, x='species', y='petal_length', palette='Set2', ax=axes[1], inner='box')
axes[1].set_title('Petal Length Distribution (Violin)', fontweight='bold')
axes[1].set_xlabel('Species')
axes[1].set_ylabel('Petal Length (cm)')

plt.suptitle('Violin Plots — Density + Spread', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.savefig('violin_plots.png', dpi=120)
plt.show()
print('Violin plots saved as violin_plots.png')
```

**Insight:** A few outliers are visible in sepal width for `setosa`. `Virginica` generally has the largest petal and sepal measurements, while `setosa` has distinctly smaller petal dimensions.

### 6.4 Correlation Heatmap (Bonus)

A heatmap shows how strongly each pair of numerical features is correlated.

```python
# Correlation heatmap of numerical features
plt.figure(figsize=(7, 5))
corr_matrix = df[features].corr()

sns.heatmap(
    corr_matrix,
    annot=True,          # Show correlation values
    fmt='.2f',           # 2 decimal places
    cmap='coolwarm',     # Color scheme
    linewidths=0.5,
    square=True,
    cbar_kws={'shrink': 0.8}
)
plt.title('Feature Correlation Heatmap', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.savefig('correlation_heatmap.png', dpi=120)
plt.show()
print('Heatmap saved as correlation_heatmap.png')
```

**Insight:** Petal length and petal width are very strongly correlated (0.96). Sepal length is also moderately correlated with petal measurements. This suggests that petal features carry the most predictive power.
